from cellpose import io, models, core, plot
import matplotlib.pyplot as plt

# load image
image = io.imread("cellpose_example/image_01.tif")

# initialize model
model = models.Cellpose(model_type='cyto3')

# run prediction
masks_pred, flows, styles, diams = model.eval(image)

# visualize result
fig, ax = plt.subplots(figsize=(8, 8))
ax.imshow(image, cmap='gray')
ax.imshow(masks_pred, alpha=0.4)
ax.set_title("Predicted masks")
ax.axis("off")

plt.show()

# Cellpose-native visualization
fig = plt.figure(figsize=(12, 5))

plot.show_segmentation(
    fig,
    image,
    masks,
    flows[0],   # flow RGB is flows[0]
